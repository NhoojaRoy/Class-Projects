```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ms = np.arange(1, 31)
svd_values = []
eigen_values = []

for m in ms:
    A = 0.1 * np.eye(m) + np.triu(np.ones((m, m)), 1)
    svd_values.append(np.linalg.svd(A, compute_uv=False)[-1])
    smallest_eigenvalue = np.linalg.eigvalsh(A.T @ A)[0]
    eigen_values.append(np.sqrt(smallest_eigenvalue))

svd_values = np.array(svd_values)
eigen_values = np.array(eigen_values)
```

```python
plt.figure(figsize=(9, 5))

plt.semilogy(ms, svd_values, "o-", label="Direct SVD", markersize=4)
plt.semilogy(
    ms, eigen_values, "s-",
    label=r"$\sqrt{\lambda_{\min}(A^T A)}$",
    markersize=4
)

plt.xlabel("Matrix size m")
plt.ylabel("Computed smallest singular value")
plt.title("Smallest singular value: two numerical methods")
plt.xticks(np.arange(1, 31, 2))
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig("singular_values.png", dpi=200)
plt.show()
```
